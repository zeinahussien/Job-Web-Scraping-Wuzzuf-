```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait


options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

```python
from urllib.parse import quote
import pandas as pd


tracks = ['Data Engineer', 'Data Science', 'AI Developer', 'Machine Learning Engineer']

driver = webdriver.Chrome(options=options)

jobs_list = []

for track in tracks: 
    print(f"========TRACK: {track} ==========")

    driver.get("https://wuzzuf.net/jobs/egypt")
    driver.find_element(By.NAME, 'q').send_keys(track)
    driver.find_element(By.XPATH, "//button[@type='submit']").click()

    time.sleep(2)
    driver.refresh()
    time.sleep(5)

    page_count = 0

    while True: 
            page_count += 1
            print('************************** PAGE NUMBER **************************', page_count)
            
            jobs = driver.find_elements(By.CLASS_NAME, 'css-pkv5jc')

            if not jobs:
                print("No jobs on page", page_count, "- done (or blocked).")
                break
            
            for i in jobs:
                job_title = i.find_element(By.CLASS_NAME, 'css-s5fwzh').text
                company = i.find_element(By.CLASS_NAME, 'css-ipsyv7').text
                location = i.find_element(By.CLASS_NAME, 'css-16x61xq').text
                job_type = i.find_element(By.CLASS_NAME, 'css-5jhz9n').text.replace('\n',' / ')

                exp = i.find_elements(By.XPATH, ".//span[contains(., 'Yrs of Exp')]")
                job_experience = exp[0].text.strip("·.").strip() if exp else None
                

                job_url = i.find_element(By.CLASS_NAME , "css-o171kl").get_attribute('href')

                jobs_list.append({
                    'Track': track,
                    'Job Title': job_title,
                    'Company': company,
                    'Location': location,
                    'Job Type': job_type,
                    'Job Experience': job_experience,
                    'Job URL': job_url
                })

            if page_count >= 40:   # safety cap
                print("Page cap reached.")
                break

            driver.get(f"https://wuzzuf.net/search/jobs?q={quote(track)}r&a=hpb&start={page_count}")
            time.sleep(5)

    df = pd.DataFrame(jobs)
    df.to_csv('wuzzuf_jobs.csv', index=False)
    print(track, "done. Total rows so far:", len(jobs_list))

    time.sleep(20)
    
driver.quit()
```
